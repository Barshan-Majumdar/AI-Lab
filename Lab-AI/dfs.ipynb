```python
import networkx as nx
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# -----------------------------
# Create Graph
# -----------------------------
G = nx.Graph()

edges = [
    ("A", "B"), ("A", "C"),
    ("B", "D"), ("B", "E"),
    ("C", "F"), ("C", "G"),
    ("E", "H")
]

G.add_edges_from(edges)

# -----------------------------
# DFS Traversal
# -----------------------------
visited = []
parent = {}
stack = ["A"]

while stack:
    node = stack.pop()

    if node not in visited:
        visited.append(node)

        # Push neighbours in reverse alphabetical order
        for neighbour in sorted(G.neighbors(node), reverse=True):
            if neighbour not in visited:
                if neighbour not in stack:
                    parent[neighbour] = node
                    stack.append(neighbour)

print("DFS Traversal Order:")
print(visited)

# -----------------------------
# Graph Layout
# -----------------------------
pos = nx.spring_layout(G, seed=5)

fig, ax = plt.subplots(figsize=(8,6))

# -----------------------------
# Animation Function
# -----------------------------
def update(frame):

    ax.clear()

    visited_now = visited[:frame+1]

    node_colors = []

    for n in G.nodes():

        if n == visited[frame]:
            node_colors.append("yellow")      # Current node

        elif n in visited_now:
            node_colors.append("red")         # Visited node

        else:
            node_colors.append("lightgray")   # Unvisited

    edge_colors = []

    for u, v in G.edges():

        if ((v in parent and parent[v] == u and v in visited_now) or
            (u in parent and parent[u] == v and u in visited_now)):
            edge_colors.append("green")
        else:
            edge_colors.append("black")

    nx.draw(
        G,
        pos,
        ax=ax,
        with_labels=True,
        node_color=node_colors,
        edge_color=edge_colors,
        edgecolors="black",
        node_size=1200,
        width=3,
        font_size=14,
        font_weight="bold"
    )

    ax.set_title(f"Current Node : {visited[frame]}")

# -----------------------------
# Animation
# -----------------------------
ani = FuncAnimation(
    fig,
    update,
    frames=len(visited),
    interval=1200,
    repeat=False
)

plt.close(fig)

HTML(ani.to_jshtml())
```
